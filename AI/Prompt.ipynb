```python
from dotenv import load_dotenv
from groq import Groq
import os

load_dotenv()

client = Groq(
    api_key=os.getenv("GROQ_API_KEY")
)

print("Connected Successfully")
```

```python
from groq import Groq

zero_shot = """Classify the Movie Review"""

movie_review1 = "Tell me the Spiderman Movie Review."

chat_completion = client.chat.completions.create(       
    messages=[
        {
            "role": "system",
            "content": zero_shot
        },
        {
            "role": "user",
            "content": movie_review1,
        }
    ],
    model="openai/gpt-oss-120b",
)

print(chat_completion.choices[0].message.content)
```

```python
from groq import Groq

one_shot = """Classify the Movie Review: 
The Movie was Bad
Sentiment Tag: Negative
Classify the Movie Review"""   

movie_review1 = "Tell me the Geetha Govindham movie review?."

chat_completion = client.chat.completions.create(
    messages=[
        {
            "role": "system",
            "content": one_shot
        },
        {
            "role": "user",
            "content": movie_review1,
        }
    ],
    model="openai/gpt-oss-120b",
)

print(chat_completion.choices[0].message.content)
```

```python
from groq import Groq

few_shot = """Classify the Movie Review: 
The Movie was Bad
Sentiment Tag: Negative
Classify the Movie Review: 
The Movie was Good
Sentiment Tag: Positive
Classify the Movie Review""" 

movie_review1 = "Tell me the Puspha Movie Review."

chat_completion = client.chat.completions.create(
    messages=[
        {
            "role": "system",
            "content": few_shot
        },
        {
            "role": "user",
            "content": movie_review1,
        }
    ],
    model="openai/gpt-oss-120b",
)

print(chat_completion.choices[0].message.content)
```

```python
from groq import Groq
cot  = """"""

prompt = """
Question:
A farmer has 15 apples. He gives 5 apples to his friend and then buys 8 more apples.
Let's think step by step.
"""    

chat_completion = client.chat.completions.create(
    messages=[
        {
            "role": "system",
            "content": cot
        },
        {
            "role": "user",
            "content": prompt,
        }
    ],
    model="llama-3.3-70b-versatile",
)

print(chat_completion.choices[0].message.content)
```
