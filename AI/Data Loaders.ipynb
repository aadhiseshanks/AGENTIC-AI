#### <i><b> RAG - Retrieval Augmented Generation
- RAG is a method where we combine LLMs with a retrieval system.
- The Retrieval System can search through vast source of external information - like documents, databases or knowledge bases whenever the LLM needs additional knowledge to give you better answers.

##### <i><b> Context Window 
- A context window is the maximum number of tokens (not words) an AI model can process or “remember” at once — for example, GPT‑4 supports up to 32,768 tokens, while some Claude models reach 1,000,000 tokens.

##### <i><b> Text Splitting get into Chunks

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
# Data Load
loader = PyPDFLoader("../Data/Agentic AI.pdf")
text = loader.load()
```

```python
len(text)
```

```python
# Chunking
data_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 120,
    chunk_overlap = 20
)
```

```python
chunk = data_splitter.split_documents(text)
```

```python
print(len(chunk))
```

##### <i><b> Embedding Models

```python
from langchain_community.vectorstores import FAISS
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
embeddings = HuggingFaceEmbeddings(
    model_name="BAAI/bge-small-en-v1.5"
)
embeddings
```

```python
vector_store = FAISS.from_documents(chunk, embeddings)
```

```python
vector_store.save_local("vector_store")
```

```python
questions = "What is Hallucination?"
```

```python
vector_store.similarity_search(questions, k=3)
```

```python
vector_store.similarity_search_with_score(questions, k=3)
```

##### <i><b> Retrieval

```python
vector_store.max_marginal_relevance_search(questions, k=3, fetch_k=10)
```
